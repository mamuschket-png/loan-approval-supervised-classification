# Daten einlesen

```python
import pandas as pd
from statsmodels.sandbox.distributions.mv_normal import BivariateNormal
from textdistance.algorithms.vector_based import Correlation

path = 'Data/Raw/'
df = pd.read_csv(path + 'loans_modified.csv')
```

Im Datensatz enthalten:
- **loan_id** – Darlehens-ID
- **gender** – Geschlecht (Male / Female)

- **dependents** – Anzahl unterhaltsberechtigter Personen

# Fixing Data Types
## Object -> String

```python
df.info()
```

```python
df.head(10).T
```

```python
df_cleaned = df.copy()
```

```python
# Eigentlich nicht nötig, denn: In pandas werden Strings als object gespeichert
df['loan_id'] = df['loan_id'].astype('string')
df['gender'] = df['gender'].astype('string')
```

```python
#print(df.select_dtypes(include='object').head())
#print()
print(df.select_dtypes(include='object').info())
```

## Column 'Dependents'

```python
print(df['dependents'].unique())
```

- es gibt also 4 Gruppen
- diese sind geordnet (2 > 1 usw.), d.h. ordinal
- one-hot encoding => ich würde mehrere Spalten erhalten und die Ordinalität verlieren
- Umwandlung in integer und 3+ zu 3 verändern => Info über "3 oder mehr" geht verloren, ich verliere alle NaNs (können in Pandas nicht in integer gespeichert werden)
- Umwandlung in float und 3+ zu 3 verändern => Info über "3 oder mehr" geht verloren, Ordinalität BLEIBT erhalten, NaNs gehen NICHT verloren

```python
df['dependents'] = df['dependents'].replace('3+', 3).astype(float)
```

## Binäre Datenspalten

```python
print(df.select_dtypes(include='object').info())
print()
print("Column married: ", df['married'].unique())
print("Column education: ", df['education'].unique())
print("Column self_employed: ", df['self_employed'].unique())
print("Column property_area: ", df['property_area'].unique())
```

```python
df = (df
  # assign = mutate
  # map um Werte zu ersetzen
  .assign(married       = lambda x: x['married'].map({'Yes': 1, 'No': 0}))
  .assign(education     = lambda x: x['education'].map({'Graduate': 1, 'Not Graduate': 0}))
  .assign(self_employed = lambda x: x['self_employed'].map({'Yes': 1, 'No': 0}))
  .assign(gender        = lambda x: x['gender'].map({'Male': 1, 'Female': 0}))
)
```

```python
print(df.select_dtypes(include='object').info())
print()
print("Column married: ", df['married'].unique())
print("Column education: ", df['education'].unique())
print("Column self_employed: ", df['self_employed'].unique())
print("Column property_area: ", df['property_area'].unique())
```

## Column property_area

- Ordinal: Rural < Semi Urban < Urban (Urbanisierungsgrad hat eine Reihenfolge)
- Nominal: Die drei Gebiete sind einfach verschiedene Kategorien ohne klare Rangfolge
- => ist es eine ordinale oder nominale Datenspalte?

- Unsicher. Im Zweifel für Nominal! => One-Hot Encoding

```python
df = pd.get_dummies(df, columns=['property_area'], drop_first=True, dtype=int)
```

zum Argument "drop_first = True":
- => mit den Eigenschaften/Säulen (1) "property_area_Semiurban" und (2) "property_area_Urban" kann auf (3) "rural" geschlossen werden
- => Eigenschaft "rural" nicht als explizit als Säule, sondern verbleibt nur implizit im Datensatz (wenn (1) und (2) False sind dann ist es rural...)
- Randnotiz: Es wird alphabetisch sortiert und die Kategorie mit Anfangsbuchstaben der am nächsten zum Anfang des Alphabets ist gelöscht

zum Argument "dtype=int":
- by default würde ich boolean dtype (true, false) erhalten
- der Datentyp macht mit manchen Modellen Probleme

### loan_amount

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x='loan_status', y='loan_amount', data=df)
plt.title("Boxplot: Loan Amount nach Loan Status")
plt.show()
```

Bivariate Analysis

```python
plt.figure(figsize=(10, 6))
sns.regplot(
    x='applicant_income',
    y='loan_amount',
    data=df,
    scatter_kws={'alpha': 0.4}, # Einstellung für Punkte
    line_kws={'color': 'red'},   # Linienfarbe
    ci=None # CI rausnehmen
)
plt.title("Lineare Regression: Applicant Income vs Loan Amount")
plt.show()
```

```python
from scipy import stats

clean = df[['applicant_income', 'loan_amount']].dropna()

slope, intercept, r_value, p_value, std_err = stats.linregress(
    clean['applicant_income'],
    clean['loan_amount']
)

print(f"R²: {r_value**2:.3f}")
print(f"p-value: {p_value:.4f}")
print(f"Slope: {slope:.4f}")
```

INTERPRETATION

- Der Zusammenhang ist statistisch signifikant (p=0.0000) aber moderat (R2 = 0.29).
- `applicant_income` erklärt nur 29% der Varianz in `loan_amount`

## Correlation matrix

```python
# Berechne Korrelationsmatrix (bilde alle paarweisen Pearson-Korrelationen)
## load_id wird gedropped, weil .corr() nichts mit Strings anfangen kann
corr_matrix = df.drop(columns='loan_id').corr(numeric_only=True)
print(corr_matrix)
print()

# Erstelle Heatmap für Korrelationsmatrix #
plt.figure(figsize=(10, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Korrelation Heatmap für Loan Daten")
plt.show()
```

INTERPRETATION

- `applicant_income` & `loan_amount` korrelieren recht stark (r=0.54) => ggf. nur eines behalten oder Feature Engineering (loan_amount / income als Ratio)
- `credit history` korreliert stark mit Zielvariable `loan status` => vmtl. auch sehr hohe feature importance im ML
- `property_area_Semiurban ` & `property_area_Urban` korrelieren recht stark (-0.52) => ergibt sich automatisch aus One-Hot-Encoding -> nicht beachten
